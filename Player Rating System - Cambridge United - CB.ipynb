# Player Attribute Ratings System

### Setup and Inputs

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
from datetime import datetime
```

```python
# Load the latest Wyscout dataset available

df = pd.read_csv("Wyscout_League_Export 2026-04-14.csv")
```

```python
# Create list of unique leagues within dataset

list_unique_leagues = list( sorted( df['League'].unique() ) )
list_unique_leagues
```

```python
# Filter settings used later in code

filter_positions = ["CB", "RCB", "LCB"]
filter_leagues = [
                  #'England Premier League 2024-25'
                  #'England Premier League 2025-26'
                  'England League One 2025-26',
                  'England League Two 2025-26'
                  ]
filter_minutes = 450

# Columns to exclude from the Z-Score Calculation
excluded_columns = ['Player', 'Team', 'Team within selected timeframe', 'Position','Age', 'Market value', 'Contract expires', 'Matches played', 'Minutes played','Birth country', 'Passport country', 'Foot', 'Height', 'Weight', 'On loan','League', 'Main Position']

# Export names
rating_export_name = "CUFC Ratings - CB.csv"
```

### Calculate Z-Scores and Export Data for Review

```python
# Create filtered df prior to applying the filters
filtered_df = df.copy()

# Apply filters to the filtered df
filtered_df = filtered_df[ filtered_df['League'].isin(filter_leagues) ]             # filter leagues
filtered_df = filtered_df[ filtered_df['Main Position'].isin(filter_positions) ]    # filter position
filtered_df = filtered_df[ filtered_df['Minutes played'] >= filter_minutes ]        # filter minutes played

filtered_df
```

```python
# Calculate the z-scores for all metric columns

z_score_df = filtered_df.copy() # create a new df that is a copy of the filtered df

columns_to_zscore = [col for col in filtered_df.columns if col not in excluded_columns] # creates a list of all columns that are not in the excluded list, i.e. a list of columns to convert into z-scores

z_score_df[columns_to_zscore] = z_score_df[columns_to_zscore].apply(zscore) # takes the metric columns subset of the filtered_df, calculates the z-score for each data point and replaces the absolute value with the z-score

# the apply function works column-wise by default - i.e. for each column, calculate the z-score for each value in the column
# then it is assign back into the original filtered_df. This only works because the output (z-scores) are the same length as the input (filtered_df)

# Invert Z-Scores for specific columns
columns_to_invert = ["Yellow cards per 90", "Red cards per 90", "Fouls per 90"]
z_score_df[columns_to_invert] = z_score_df[columns_to_invert] * -1

z_score_df
```

```python
# Function to normalize the data to a [0, 1] range
def NormalizeData(data):
    # Subtract the minimum value from each element and divide by the range
    return (data - np.min(data)) / (np.max(data) - np.min(data))

# Iterate over the specified columns (columns_to_zscore) to normalize them
for column in columns_to_zscore:
    # Apply the NormalizeData function to scale the column values
    z_score_df[column] = NormalizeData(z_score_df[column])

z_score_df
```

### Attribute Definition and Calculation

```python
# Create new rating columns in a batch
new_columns = {
                # Template line: 'REPLACE Rating': z_score_df['REPLACE'] * 0,

                # Defending - 40%
                'Defensive duels won % Rating':               z_score_df['Defensive duels won, %']                 * 0.35,
                'PAdj Interceptions Rating':                  z_score_df['PAdj Interceptions']                     * 0.20,
                'PAdj Sliding tackles Rating':                z_score_df['PAdj Sliding tackles']                   * 0.20,
                'Successful defensive actions per 90 Rating': z_score_df['Successful defensive actions per 90']    * 0.125,
                'Shots blocked per 90 Rating':                z_score_df['Shots blocked per 90']                   * 0.125,
                
                # Aerial - 30%
                'Aerial duels won % Rating':   z_score_df['Aerial duels won, %'] * 0.60,
                'Aerial duels per 90 Rating':  z_score_df['Aerial duels per 90'] * 0.40,
                
                # Pass Retention - 10%
                'Accurate short passes % Rating':   z_score_df['Accurate short / medium passes, %'] * 0.50,
                'Short passes per 90 Rating':       z_score_df['Short / medium passes per 90']      * 0.30,
                'Received passes per 90 Rating':    z_score_df['Received passes per 90']            * 0.20,

                # Pass Progression - 5%
                'Progressive passes per 90 Rating':        z_score_df['Progressive passes per 90']         * 0.40,
                'Accurate progressive passes % Rating':    z_score_df['Accurate progressive passes, %']    * 0.60,
                
                # Discipline 15%
                'Fouls per 90 Rating':         z_score_df['Fouls per 90']     * 0.50,
                'Yellows per 90 Rating':  z_score_df['Yellow cards per 90']   * 0.30,
                'Reds per 90 Rating':     z_score_df['Red cards per 90']      * 0.20
                }

# Create the new attribute columns using the above ratings columns
# Template line: new_columns['REPLACE Att'] = new_columns['REPLACE'] + new_columns['REPLACE'] + new_columns['REPLACE'] + new_columns['REPLACE']
new_columns['Defending Att'] = new_columns['Defensive duels won % Rating'] + new_columns['PAdj Interceptions Rating'] + new_columns['PAdj Sliding tackles Rating'] + new_columns['Successful defensive actions per 90 Rating'] + new_columns['Shots blocked per 90 Rating']
new_columns['Aerial Att']    = new_columns['Aerial duels won % Rating'] + new_columns['Aerial duels per 90 Rating']
new_columns['Pass Retention Att']   = new_columns['Accurate short passes % Rating'] + new_columns['Short passes per 90 Rating'] + new_columns['Received passes per 90 Rating']
new_columns['Pass Progression Att'] = new_columns['Progressive passes per 90 Rating'] + new_columns['Accurate progressive passes % Rating']
new_columns['Discipline Att']     = new_columns['Fouls per 90 Rating'] + new_columns['Yellows per 90 Rating'] + new_columns['Reds per 90 Rating']

# Add all new columns to the DataFrame at once
z_score_df = pd.concat([z_score_df, pd.DataFrame(new_columns)], axis=1)

# Create a fresh copy of the df to avoid issues if re-running cells
z_score_df = z_score_df.copy()
z_score_df
```

```python
# Create new df containing absolute values to append onto final ratings df later
absolute_columns = [# For join
                    'Player',
                    'Team within selected timeframe',
                    
                    # Defending
                    'Defensive duels won, %',
                    'PAdj Interceptions',
                    'PAdj Sliding tackles',
                    'Successful defensive actions per 90',
                    'Shots blocked per 90',
                    
                    # Aerial
                    'Aerial duels won, %',
                    'Aerial duels per 90',
                    
                    # Pass Retention
                    'Accurate short / medium passes, %',
                    'Short / medium passes per 90',
                    'Received passes per 90',
                    
                    # Pass Progression
                    'Progressive passes per 90',
                    'Accurate progressive passes, %',
                    
                    # Discipline
                    'Fouls per 90',
                    'Yellow cards per 90',
                    'Red cards per 90',

                    # Set Pieces
                    'Free kicks per 90',
                    'Direct free kicks per 90',
                    'Direct free kicks on target, %',
                    'Corners per 90',
                    
                    ]
absolute_values = filtered_df[absolute_columns].copy()

absolute_values
```

```python
# Creating the new DataFrame 'ratings' with the specified columns
ratings_columns = [
                    "Player", "Team", "Team within selected timeframe", "Main Position", "Position", "League",
                    "Age", "Market value", "Contract expires", "On loan", "Minutes played",
                    
                    # New Att columns created earlier:
                    "Defending Att", "Aerial Att", "Pass Retention Att", "Pass Progression Att", "Discipline Att"
                    ]

ratings = z_score_df[ratings_columns].copy()

ratings
```

```python
# Normalise the data again after we have created attributes

# Ensure a copy of the DataFrame to prevent unintended changes to the original data
norm_ratings = ratings.copy()

# Select columns that end with the word "Att"
columns_to_normalize = [col for col in norm_ratings.columns if col.endswith('Att')]

# Iterate over the selected columns and apply normalization
for column in columns_to_normalize:

    # Apply the NormalizeData function to scale the column
    norm_ratings[column] = NormalizeData(norm_ratings[column])

norm_ratings
```

### Weighting the Attributes to Create an Overall Rating

```python
# Weight the overall attributes in terms of importance 
weighted_ratings = norm_ratings.copy()

#Weight the overall ratings
weighted_ratings['Defending Att'] =        weighted_ratings['Defending Att']        * 0.40
weighted_ratings['Aerial Att'] =           weighted_ratings['Aerial Att']           * 0.30
weighted_ratings['Pass Retention Att'] =   weighted_ratings['Pass Retention Att']   * 0.10
weighted_ratings['Pass Progression Att'] = weighted_ratings['Pass Progression Att'] * 0.05
weighted_ratings['Discipline Att'] =       weighted_ratings['Discipline Att']       * 0.15

weighted_ratings['CB Rating'] = ( weighted_ratings['Defending Att'] 
                                + weighted_ratings['Aerial Att'] 
                                + weighted_ratings['Pass Retention Att']
                                + weighted_ratings['Pass Progression Att'] 
                                + weighted_ratings['Discipline Att'] 
                                )

weighted_ratings
```

```python
# Final time to normalise, this time for the overall rating 

# Ensure a copy of the DataFrame to prevent unintended changes
final_ratings = weighted_ratings.copy()

# Select the attribute and rating columns in the DataFrame
columns_to_normalize = [col for col in final_ratings.columns if col.endswith('Att') or col.endswith('Rating')]

# Normalise the attribute and rating columns
for column in columns_to_normalize:
    final_ratings[column] = NormalizeData(final_ratings[column]) # function defined earlier

# Multiply columns ending with 'Att' or 'Rating' by 100 and round to 2 decimal places
final_ratings[columns_to_normalize] = (final_ratings[columns_to_normalize] * 100).round(2)

# Sort the DataFrame by the last column in descending order
rating_column = [col for col in final_ratings.columns if col.endswith('Rating')]
final_ratings.sort_values(by=rating_column, ascending=False, inplace=True)

# Merge the absolute values
final_ratings = final_ratings.merge(absolute_values,on=['Player','Team within selected timeframe'], how='left')

# Reset the index of the sorted DataFrame to ensure a clean, sequential index
final_ratings.reset_index(drop=True, inplace=True)

# Export the ratings to CSV
final_ratings.to_csv(rating_export_name, index=False)
print(f"Ratings exported at {datetime.now().strftime('%d-%m-%Y %H:%M')}")

final_ratings
```
